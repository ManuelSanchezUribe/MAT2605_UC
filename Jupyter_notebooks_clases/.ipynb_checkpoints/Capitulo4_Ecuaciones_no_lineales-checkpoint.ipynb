# Metodo de biseccion


```python
def biseccion(f, a, b, TOL=1e-12, MAXITER=100, full_output=False):
    """
    Parámetros:
        f (function): Función no lineal f(x).
        a (float): Límite inferior del intervalo.
        b (float): Límite superior del intervalo.
        TOL (float, opcional): Tolerancia para el criterio de parada.
        MAXITER (int, opcional): Número máximo de iteraciones.
        full_output (bool, opcional): Si es True, retorna la historia de iteraciones.

    Retorna:
        Si full_output == False:
            c (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
        Si full_output == True:
            c (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
            history (dict): Diccionario con intervalos, puntos medios y evaluaciones.
    """
    if f(a) * f(b) >= 0:
        raise ValueError("f(a) y f(b) deben tener signos opuestos.")

    i = 0
    intervals, clist = [], []
    f_alist, f_blist, f_clist = [], [], []
    
    while (b - a) / 2 > TOL and i < MAXITER:
        c = (a + b) / 2.0
        fa, fb, fc = f(a), f(b), f(c)
        
        if full_output:
            intervals.append((a, b))
            clist.append(c)
            f_alist.append(fa)
            f_blist.append(fb)
            f_clist.append(fc)

        if abs(fc) < TOL:
            break
        elif fa * fc < 0:
            b = c
        else:
            a = c
        i += 1

    if not full_output:
        return c, i
    else:
        history = {
            'intervals': intervals,
            'clist': clist,
            'f_alist': f_alist,
            'f_blist': f_blist,
            'f_clist': f_clist
        }
        return c, i, history
```

```python
a = 0.0
b = 3.0
fun = lambda x: (x-0.5)*(x+3.5)
# Ejecutar el algoritmo
c, n = biseccion(fun, a, b)
print(f"Raíz: {c:.12f} en {n} iteraciones.")
```

```python
def ilustrar_biseccion(k):
    ak, bk = intervals[k]
    ck = clist[k]
    fak, fbk, fck = f_alist[k], f_blist[k], f_clist[k]

    fig, ax = plt.subplots(figsize=(10, 6))

    # --- Zoom adaptativo con 20% de margen ---
    ancho = bk - ak
    lim_izq, lim_der = ak - 0.20 * ancho, bk + 0.20 * ancho
    xplot = np.linspace(lim_izq, lim_der, 500)
    yplot = fun(xplot)

    # Curva f(x) y eje y = 0
    ax.plot(xplot, yplot, label="$f(x)$", color="#1f77b4", lw=2.5)
    ax.axhline(0, color="black", linestyle="--", linewidth=1.0)

    # --- Marcadores de evaluación f(a), f(b) y f(c) sobre la curva ---
    ax.plot(ak, fak, 's', color='#d62728', markersize=9, label=f"$f(a_{{{k}}}) = {fak:.3f}$")
    ax.plot(bk, fbk, 's', color='#9467bd', markersize=9, label=f"$f(b_{{{k}}}) = {fbk:.3f}$")
    ax.plot(ck, fck, 'o', color='#2ca02c', markersize=9, label=f"$f(c_{{{k}}}) = {fck:.3f}$")

    # Proyecciones verticales desde la curva hacia el eje X
    ax.vlines(x=[ak, bk, ck], ymin=0, ymax=[fak, fbk, fck], colors=['#d62728', '#9467bd', '#2ca02c'], linestyles=':', lw=1.5)

    # Marcadores en el eje X
    ax.plot(ck, 0, 'X', color='#2ca02c', markersize=9, label=f"Punto medio $c_{{{k}}} = {ck:.5f}$")

    # --- Determinar y resaltar el Siguiente Intervalo ---
    if fak * fck < 0:
        next_a, next_b = ak, ck
        razon = "Como $f(a_k) \cdot f(c_k) < 0$, la raíz está en $[a_k, c_k]$"
    else:
        next_a, next_b = ck, bk
        razon = "Como $f(a_k) \cdot f(c_k) > 0$, la raíz está en $[c_k, b_k]$"

    # Resaltar el área del intervalo actual e intervalo siguiente
    ax.plot([ak, bk], [0, 0], 'o-', color='#ff7f0e', lw=4, label=f"Intervalo actual $I_{{{k}}}$")
    ax.axvspan(next_a, next_b, color='#2ca02c', alpha=0.18, label=f"Siguiente intervalo $I_{{{k+1}}}$")

    # Ajuste de límites del gráfico
    ax.set_xlim([lim_izq, lim_der])
    y_min, y_max = min(yplot), max(yplot)
    alto_y = y_max - y_min
    ax.set_ylim([y_min - 0.15 * alto_y, y_max + 0.15 * alto_y])

    # Títulos e información
    ax.set_title(f"Iteración {k + 1} de {MAX_ITER_VIS}\n{razon}", fontsize=12, fontweight='bold', pad=12)
    ax.set_xlabel("$x$", fontsize=11)
    ax.set_ylabel("$f(x)$", fontsize=11)
    ax.grid(True, which='both', alpha=0.3)
    ax.legend(loc="upper left", framealpha=0.9)

    plt.tight_layout()
    plt.show()

# Slider restringido a las primeras N iteraciones
slider = IntSlider(min=0, max=MAX_ITER_VIS - 1, step=1, value=0, description='Iteración:')
interact(ilustrar_biseccion, k=slider);
```

# Metodo de regula falsi

```python
def regula_falsi(f, a, b, TOL=1e-12, MAXITER=100, full_output=False):
    """
    Algoritmo de Regula Falsi (Falsa Posición).
    
    Parámetros:
        f (function): Función no lineal f(x).
        a (float): Límite inferior del intervalo.
        b (float): Límite superior del intervalo.
        TOL (float, opcional): Tolerancia para el criterio de parada.
        MAXITER (int, opcional): Número máximo de iteraciones.
        full_output (bool, opcional): Si es True, retorna el historial de iteraciones.

    Retorna:
        Si full_output == False:
            c (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
        Si full_output == True:
            c (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
            history (dict): Diccionario con intervalos, raíces calculadas y evaluaciones.
    """
    if f(a) * f(b) >= 0:
        raise ValueError("f(a) y f(b) deben tener signos opuestos.")

    i = 0
    intervals, clist = [], []
    f_alist, f_blist, f_clist = [], [], []

    while i < MAXITER:
        fa, fb = f(a), f(b)
        
        # Fórmula de Regula Falsi (intersección de la recta secante con el eje X)
        c = a - fa * (b - a) / (fb - fa)
        fc = f(c)

        if full_output:
            intervals.append((a, b))
            clist.append(c)
            f_alist.append(fa)
            f_blist.append(fb)
            f_clist.append(fc)

        # Criterio de parada por tolerancia en f(c) o en la longitud del paso
        if abs(fc) < TOL:
            break

        if fa * fc < 0:
            b = c
        else:
            a = c

        i += 1

    if not full_output:
        return c, i
    else:
        history = {
            'intervals': intervals,
            'clist': clist,
            'f_alist': f_alist,
            'f_blist': f_blist,
            'f_clist': f_clist
        }
        return c, i, history


# 3. Resolver y obtener el historial completo
c_sol, n_iter, hist = regula_falsi(fun, a_init, b_init, full_output=True)

intervals = hist['intervals']
clist = hist['clist']
f_alist, f_blist, f_clist = hist['f_alist'], hist['f_blist'], hist['f_clist']

# Definir la cantidad de iteraciones a ilustrar (ejemplo: las primeras 6)
MAX_ITER_VIS = min(6, len(clist))


# 4. Función de visualización didáctica para Regula Falsi
def ilustrar_regula_falsi(k):
    ak, bk = intervals[k]
    ck = clist[k]
    fak, fbk, fck = f_alist[k], f_blist[k], f_clist[k]

    fig, ax = plt.subplots(figsize=(10, 6))

    # --- Zoom adaptativo enfocado en el intervalo actual con 20% de margen ---
    ancho = bk - ak
    lim_izq, lim_der = ak - 0.20 * ancho, bk + 0.20 * ancho
    xplot = np.linspace(lim_izq, lim_der, 500)
    yplot = fun(xplot)

    # Curva f(x) y eje y = 0
    ax.plot(xplot, yplot, label="$f(x)$", color="#1f77b4", lw=2.5)
    ax.axhline(0, color="black", linestyle="--", linewidth=1.0)

    # --- Recta Secante entre (ak, f(ak)) y (bk, f(bk)) ---
    ax.plot([ak, bk], [fak, fbk], color='black', linestyle='--', lw=1.8, label="Recta secante")

    # --- Marcadores de evaluación sobre la curva ---
    ax.plot(ak, fak, 's', color='#d62728', markersize=9, label=f"$f(a_{{{k}}}) = {fak:.3f}$")
    ax.plot(bk, fbk, 's', color='#9467bd', markersize=9, label=f"$f(b_{{{k}}}) = {fbk:.3f}$")
    ax.plot(ck, fck, 'o', color='#2ca02c', markersize=9, label=f"$f(c_{{{k}}}) = {fck:.3f}$")

    # Proyecciones verticales desde la curva hacia el eje X
    ax.vlines(x=[ak, bk, ck], ymin=0, ymax=[fak, fbk, fck], colors=['#d62728', '#9467bd', '#2ca02c'], linestyles=':', lw=1.5)

    # Intersección en el eje X
    ax.plot(ck, 0, 'X', color='#2ca02c', markersize=9, label=f"Intersección secante $c_{{{k}}} = {ck:.5f}$")

    # --- Determinar y resaltar el Siguiente Intervalo ---
    if fak * fck < 0:
        next_a, next_b = ak, ck
        razon = f"Como $f(a_k) \\cdot f(c_k) < 0$, la raíz está en $[a_k, c_k]$"
    else:
        next_a, next_b = ck, bk
        razon = f"Como $f(a_k) \\cdot f(c_k) > 0$, la raíz está en $[c_k, b_k]$"

    # Intervalo actual y siguiente subintervalo
    ax.plot([ak, bk], [0, 0], 'o-', color='#ff7f0e', lw=4, label=f"Intervalo actual $I_{{{k}}}$")
    ax.axvspan(next_a, next_b, color='#2ca02c', alpha=0.18, label=f"Siguiente $I_{{{k+1}}}$")

    # Ajustar límites Y
    y_min, y_max = min(yplot), max(yplot)
    alto_y = y_max - y_min
    ax.set_xlim([lim_izq, lim_der])
    ax.set_ylim([y_min - 0.15 * alto_y, y_max + 0.15 * alto_y])

    # Títulos y detalles
    ax.set_title(f"Regula Falsi — Iteración {k + 1} de {MAX_ITER_VIS}\n{razon}", fontsize=11, fontweight='bold', pad=12)
    ax.set_xlabel("$x$", fontsize=11)
    ax.set_ylabel("$f(x)$", fontsize=11)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left", framealpha=0.9)

    plt.tight_layout()
    plt.show()

# Controles interactivos con deslizador
slider = IntSlider(min=0, max=MAX_ITER_VIS - 1, step=1, value=0, description='Iteración:')
interact(ilustrar_regula_falsi, k=slider);
```

# Metodo de Newton-Raphson

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# 1. Definición del problema
fun = lambda x: (x - 0.5) * (x + 3.5)
dfun = lambda x: 2 * x + 3.0
x0_init = 1.5

# 2. Algoritmo de Newton-Raphson con opción 'full_output'
def newton_raphson(f, df, x0, TOL=1e-12, MAXITER=100, full_output=False):
    """
    Algoritmo de Newton-Raphson.

    Parámetros:
        f (function): Función no lineal f(x).
        df (function): Derivada de la función f'(x).
        x0 (float): Valor inicial (punto de partida).
        TOL (float, opcional): Tolerancia para el criterio de parada.
        MAXITER (int, opcional): Número máximo de iteraciones.
        full_output (bool, opcional): Si es True, retorna el historial de iteraciones.

    Retorna:
        Si full_output == False:
            x (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
        Si full_output == True:
            x (float): Raíz aproximada.
            i (int): Número de iteraciones realizadas.
            history (dict): Diccionario con aproximaciones, evaluaciones y derivadas.
    """
    i = 0
    xk_list = [x0]
    f_xlist = [f(x0)]
    df_xlist = [df(x0)]

    x = x0
    while i < MAXITER:
        fx = f(x)
        dfx = df(x)

        if abs(dfx) < 1e-15:
            raise ZeroDivisionError("La derivada f'(x) es extremadamente cercana a cero.")

        # Paso de Newton-Raphson
        x_next = x - fx / dfx

        if full_output:
            xk_list.append(x_next)
            f_xlist.append(f(x_next))
            df_xlist.append(df(x_next))

        if abs(f(x_next)) < TOL or abs(x_next - x) < TOL:
            break

        x = x_next
        i += 1

    if not full_output:
        return x_next, i + 1
    else:
        history = {
            'xk_list': xk_list,
            'f_xlist': f_xlist,
            'df_xlist': df_xlist
        }
        return x_next, i + 1, history


# 3. Resolver y obtener historial completo
x_sol, n_iter, hist = newton_raphson(fun, dfun, x0_init, full_output=True)

xk_list = hist['xk_list']
f_xlist = hist['f_xlist']
df_xlist = hist['df_xlist']

# Definir la cantidad de iteraciones a ilustrar (ejemplo: las primeras 5)
MAX_ITER_VIS = min(5, len(xk_list) - 1)


# 4. Función de visualización didáctica para Newton-Raphson
def ilustrar_newton(k):
    xk = xk_list[k]
    xk_next = xk_list[k + 1]
    fxk = f_xlist[k]
    dfxk = df_xlist[k]

    fig, ax = plt.subplots(figsize=(10, 6))

    # --- Zoom adaptativo enfocado entre x_k y x_{k+1} ---
    ancho = abs(xk_next - xk)
    if ancho < 1e-6:
        ancho = 1e-6 # Evitar división por cero o zoom plano
        
    lim_izq = min(xk, xk_next) - 0.35 * ancho
    lim_der = max(xk, xk_next) + 0.35 * ancho
    xplot = np.linspace(lim_izq, lim_der, 500)
    yplot = fun(xplot)

    # Recta tangente T(x) = f(xk) + f'(xk)*(x - xk)
    tangente = lambda x: fxk + dfxk * (x - xk)

    # Curva f(x) y eje y = 0
    ax.plot(xplot, yplot, label="$f(x)$", color="#1f77b4", lw=2.5)
    ax.axhline(0, color="black", linestyle="--", linewidth=1.0)

    # --- Recta Tangente ---
    ax.plot(xplot, tangente(xplot), color='black', linestyle='--', lw=1.8,
            label=f"Tangente $T_{{{k}}}(x)$ (pendiente $= {dfxk:.2f}$)")

    # --- Marcadores de evaluación sobre la curva ---
    ax.plot(xk, fxk, 'o', color='#d62728', markersize=9, label=f"$f(x_{{{k}}}) = {fxk:.4f}$")
    ax.plot(xk_next, fun(xk_next), 'o', color='#2ca02c', markersize=9, label=f"$f(x_{{{k+1}}}) = {fun(xk_next):.2e}$")

    # Proyección vertical desde la curva hacia el eje X en x_k
    ax.vlines(x=xk, ymin=0, ymax=fxk, colors='#d62728', linestyles=':', lw=1.5)
    ax.vlines(x=xk_next, ymin=0, ymax=fun(xk_next), colors='#2ca02c', linestyles=':', lw=1.5)

    # Puntos sobre el eje X
    ax.plot(xk, 0, 's', color='#d62728', markersize=8, label=f"$x_{{{k}}} = {xk:.6f}$")
    ax.plot(xk_next, 0, 'X', color='#2ca02c', markersize=9, label=f"$x_{{{k+1}}} = {xk_next:.6f}$")

    # Ajustar límites de los ejes
    y_vals = np.concatenate([yplot, tangente(xplot)])
    y_min, y_max = min(y_vals), max(y_vals)
    alto_y = y_max - y_min
    ax.set_xlim([lim_izq, lim_der])
    ax.set_ylim([y_min - 0.15 * alto_y, y_max + 0.15 * alto_y])

    # Títulos e información
    razon = f"Siguiente aproximación: $x_{{{k+1}}} = x_{{{k}}} - \\frac{{f(x_{{{k}}})}}{{f'(x_{{{k}}})}} = {xk_next:.8f}$"
    ax.set_title(f"Newton-Raphson — Iteración {k + 1} de {MAX_ITER_VIS}\n{razon}", fontsize=11, fontweight='bold', pad=12)
    ax.set_xlabel("$x$", fontsize=11)
    ax.set_ylabel("$f(x)$", fontsize=11)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left", framealpha=0.9)

    plt.tight_layout()
    plt.show()

# Controles interactivos con deslizador
slider = IntSlider(min=0, max=MAX_ITER_VIS - 1, step=1, value=0, description='Iteración:')
interact(ilustrar_newton, k=slider);
```

# Iteracion de punto fijo

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# 1. Definición del problema: x = g(x)
# f(x) = x^2 + 3x - 1.75 = 0  =>  x*(x + 3) = 1.75  =>  x = 1.75 / (x + 3)
gfun = lambda x: 1.75 / (x + 3.0)
x0_init = 1.5

# 2. Algoritmo de Punto Fijo con opción 'full_output'
def punto_fijo(g, x0, TOL=1e-12, MAXITER=100, full_output=False):
    """
    Algoritmo de Iteración de Punto Fijo: x_{k+1} = g(x_k).

    Parámetros:
        g (function): Función de iteración g(x).
        x0 (float): Valor inicial.
        TOL (float, opcional): Tolerancia para el criterio de parada |x_{k+1} - x_k|.
        MAXITER (int, opcional): Número máximo de iteraciones.
        full_output (bool, opcional): Si es True, retorna el historial de iteraciones.

    Retorna:
        Si full_output == False:
            x (float): Punto fijo/raíz aproximada.
            i (int): Número de iteraciones realizadas.
        Si full_output == True:
            x (float): Punto fijo/raíz aproximada.
            i (int): Número de iteraciones realizadas.
            history (dict): Historial con las aproximaciones y evaluaciones.
    """
    i = 0
    xk_list = [x0]
    gx_list = [g(x0)]

    x = x0
    while i < MAXITER:
        gx = g(x)

        if full_output and i > 0:
            xk_list.append(x)
            gx_list.append(gx)

        if abs(gx - x) < TOL:
            break

        x = gx
        i += 1

    if not full_output:
        return x, i
    else:
        # Asegurar que el último paso quede guardado
        if len(xk_list) <= i:
            xk_list.append(x)
            gx_list.append(g(x))

        history = {
            'xk_list': xk_list,
            'gx_list': gx_list
        }
        return x, i, history


tol = 10**(-8)
a = -1
b = 4
x0 = 0
fun = lambda x: x**2-3*x-1
gfun = lambda x: (x**2-1)/3
x_sol, n_iter, hist = punto_fijo(gfun, x0, full_output=True)

xk_list = hist['xk_list']
gx_list = hist['gx_list']

# Cantidad de iteraciones a ilustrar (ejemplo: las primeras 6)
MAX_ITER_VIS = min(6, len(xk_list) - 1)


# 4. Función de visualización didáctica (Diagrama de Telaraña / Cobweb Plot)
def ilustrar_punto_fijo(k):
    xk = xk_list[k]
    gxk = gx_list[k]
    xk_next = gxk  # x_{k+1} = g(x_k)

    fig, ax = plt.subplots(figsize=(10, 6))

    # --- Zoom adaptativo enfocando la trayectoria ---
    puntos_x = xk_list[:k+2]
    puntos_y = gx_list[:k+2]
    
    ancho = max(puntos_x) - min(puntos_x)
    if ancho < 1e-5:
        ancho = 1e-5

    lim_izq = min(puntos_x) - 0.25 * ancho
    lim_der = max(puntos_x) + 0.25 * ancho
    
    xplot = np.linspace(lim_izq, lim_der, 500)
    yplot = gfun(xplot)

    # 1. Curva y = g(x) y recta Identidad y = x
    ax.plot(xplot, yplot, label="$y = g(x)$", color="#1f77b4", lw=2.5)
    ax.plot(xplot, xplot, label="$y = x$", color="black", linestyle="--", lw=1.5)

    # 2. Trayectoria de Telaraña (Cobweb Plot) hasta la iteración k
    for j in range(k + 1):
        xj = xk_list[j]
        gxj = gx_list[j]

        # Paso 1: Tramo vertical desde (x_j, x_j) o (x_j, 0) hasta (x_j, g(x_j))
        if j == 0:
            ax.vlines(x=xj, ymin=0, ymax=gxj, colors='#d62728', linestyles=':', lw=1.5)
        else:
            prev_g = gx_list[j-1]
            ax.vlines(x=xj, ymin=prev_g, ymax=gxj, colors='#ff7f0e', linestyles='--', lw=1.5)

        # Paso 2: Tramo horizontal desde (x_j, g(x_j)) hasta la recta y = x en (g(x_j), g(x_j))
        ax.hlines(y=gxj, xmin=xj, xmax=gxj, colors='#2ca02c', linestyles='--', lw=1.5)

    # 3. Marcadores de la iteración actual k
    ax.plot(xk, gxk, 'o', color='#d62728', markersize=9, label=f"$(x_{{{k}}}, g(x_{{{k}}})) = ({xk:.4f}, {gxk:.4f})$")
    ax.plot(xk_next, xk_next, 'X', color='#2ca02c', markersize=9, label=f"Proyección $x_{{{k+1}}} = {xk_next:.6f}$")

    # Proyección al eje X
    ax.vlines(x=xk_next, ymin=0, ymax=xk_next, colors='#2ca02c', linestyles=':', lw=1.2)
    ax.plot(xk, 0, 's', color='#d62728', markersize=7, label=f"$x_{{{k}}} = {xk:.6f}$")

    # Ajuste de límites del gráfico
    y_min, y_max = min(np.concatenate([yplot, xplot])), max(np.concatenate([yplot, xplot]))
    alto_y = y_max - y_min
    ax.set_xlim([lim_izq, lim_der])
    ax.set_ylim([y_min - 0.10 * alto_y, y_max + 0.10 * alto_y])

    # Títulos e información
    razon = f"Evaluación: $g(x_{{{k}}}) = {gxk:.6f} \\longrightarrow x_{{{k+1}}} = {xk_next:.6f}$"
    ax.set_title(f"Punto Fijo — Iteración {k + 1} de {MAX_ITER_VIS}\n{razon}", fontsize=11, fontweight='bold', pad=12)
    ax.set_xlabel("$x$", fontsize=11)
    ax.set_ylabel("$y$", fontsize=11)
    ax.grid(True, alpha=0.3)
    ax.legend(loc="upper left", framealpha=0.9)

    plt.tight_layout()
    plt.show()

# Controles interactivos con deslizador
slider = IntSlider(min=0, max=MAX_ITER_VIS - 1, step=1, value=0, description='Iteración:')
interact(ilustrar_punto_fijo, k=slider);
```

# Metodo de la secante

```python
def secante(fun,x0,x1,n):
    xk = []
    xk.append(x0)
    xk.append(x1)
    k = 1
    while k<n and abs(fun(xk[k]))>tol:
        x = xk[k] - fun(xk[k])*(xk[k]-xk[k-1])/(fun(xk[k])-fun(xk[k-1]))
        xk.append(x)
        k+=1
    return xk
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Desactivar advertencias de NumPy para divisiones por cero o raíces de negativos
np.seterr(all='ignore') 

# ==============================================================================
# 1. Definición de Funciones g(x) y sus Derivadas g'(x)
# ==============================================================================

# Lista de funciones g(x)
g_fun = [
    lambda x: x - x**3 - 4*x**2 + 10,
    lambda x: np.sqrt(10/x - 4*x),
    lambda x: 0.5 * np.sqrt(10 - x**3),
    lambda x: np.sqrt(10 / (4 + x)),
    lambda x: x - (x**3 + 4*x**2 - 10) / (3*x**2 + 8*x)
]

# Lista de derivadas g'(x) - Calculadas a mano
g_prime_fun = [
    lambda x: 1 - 3*x**2 - 8*x, # g1'(x)
    lambda x: 0.5 / np.sqrt(10/x - 4*x) * (-10/x**2 - 4), # g2'(x)
    lambda x: 0.5 / (2 * np.sqrt(10 - x**3)) * (-3 * x**2), # g3'(x)
    lambda x: 0.5 / np.sqrt(10 / (4 + x)) * (-10 / (4 + x)**2), # g4'(x)
    
    # g5'(x) - Derivada de la fórmula de Newton-Raphson (f''f / (f')^2)
    lambda x: ((3*x**2 + 8*x) * (3*x**2 + 8*x) - (x**3 + 4*x**2 - 10) * (6*x + 8)) / (3*x**2 + 8*x)**2
]

# Nombres y estilos
colores = ['#0176DE', '#00AA00', '#FF6B00', '#F24F4F', '#455A64']
nombres = ['$g_1(x)$', '$g_2(x)$', '$g_3(x)$', '$g_4(x)$', '$g_5(x)$']
styles = ['-', '--'] 

# ==============================================================================
# 2. Setup Plotting Domain
# ==============================================================================
intervalo = [1.0, 2.0]
X = np.linspace(intervalo[0], intervalo[1], 500)

# ==============================================================================
# 3. Plotting Loop
# ==============================================================================
for i in range(len(g_fun)):
    
    # Calcular Y values
    Y_g = g_fun[i](X)
    Y_gp = g_prime_fun[i](X)
    
    # --- Robust Error Handling ---
    # 1. Check for complex numbers (result of sqrt of negative) and take the real part
    if np.iscomplexobj(Y_g): Y_g = Y_g.real
    if np.iscomplexobj(Y_gp): Y_gp = Y_gp.real

    # 2. Mask invalid values (NaN, Inf) which arise from domain errors (key to avoid crashes)
    Y_g = np.where(np.isfinite(Y_g), Y_g, np.nan)
    Y_gp = np.where(np.isfinite(Y_gp), Y_gp, np.nan)
    
    # Create the figure
    fig, ax = plt.subplots(figsize=(6, 4))
    
    # Plot g(x) and g'(x)
    ax.plot(X, Y_g, styles[0], color=colores[i], label=f'Función {nombres[i]}')
    ax.plot(X, Y_gp, styles[1], color=colores[i], label=f'Derivada $g\'_{i+1}(x)$')
    
    # Reference lines for convergence criteria (|g'(x)| < 1)
    ax.axhline(0, color='gray', linestyle=':', linewidth=0.8) 
    ax.axhline(1, color='red', linestyle=':', linewidth=1.2)   
    ax.axhline(-1, color='red', linestyle=':', linewidth=1.2)  
    ax.axvline(1, color='gray', linestyle=':', linewidth=0.8)
    ax.axvline(2, color='gray', linestyle=':', linewidth=0.8)
    
    # Labels and title
    ax.set_title(f'Gráfica de {nombres[i]} y su Derivada $g\'_{i+1}(x)$ en [1, 2]')
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')
    ax.legend(loc='best')
    ax.grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()

# Revert NumPy error settings
np.seterr(all='warn')
```

```python
def punto_fijo(g, p0, tol=1e-6, max_iter=50):
    p = p0
    for i in range(1, max_iter + 1):
        p_new = g(p)
        if np.abs(p_new - p) < tol:
            return p_new, i
        p = p_new
    return np.nan, max_iter
```

```python
import pandas as pd
g_fun = [
    lambda x: x - x**3 - 4*x**2 + 10,
    lambda x: np.sqrt(abs(10/x - 4*x)),
    lambda x: 0.5 * np.sqrt(10 - x**3),
    lambda x: np.sqrt(10 / (4 + x)),
    lambda x: x - (x**3 + 4*x**2 - 10) / (3*x**2 + 8*x)
]

# Inicialización de parámetros
p0_inicial = 1.5
tolerancia = 1e-6
max_iteraciones = 50

resultados = []

# Iterar sobre todas las funciones
for i, g in enumerate(g_fun):
    nombre = f'g_{i+1}(x)'
    
    try:
        # 1. Intentar calcular el punto fijo
        raiz, iteraciones = punto_fijo(g, p0_inicial, tolerancia, max_iteraciones)
        
        # 2. Calcular el valor de f(x) = g(x) - x para la raíz encontrada
        # Usamos la función original f(x) = x^3 + 4x^2 - 10, que es la que se está resolviendo
        # (Nota: f(x) = 0 debe ser el resultado)
        f_x_minus_x = raiz**3 + 4*raiz**2 - 10 if not np.isnan(raiz) else np.nan
        
        estado = "Convergencia" if not np.isnan(raiz) else "No Converge"
        
        resultados.append({
            'Función': nombre,
            'Estado': estado,
            'Iteraciones': iteraciones if not np.isnan(raiz) else '>',
            'Raíz Aproximada': f'{raiz:.6f}' if not np.isnan(raiz) else 'N/A',
            '|g(x)-x| (Error Final)': f'{np.abs(f_x_minus_x):.2e}' if not np.isnan(raiz) else 'N/A'
        })
        
    except Exception as e:
        # Captura errores de dominio (ej. raíz de un negativo)
        resultados.append({
            'Función': nombre,
            'Estado': "Error de Dominio",
            'Iteraciones': 'N/A',
            'Raíz Aproximada': 'N/A',
            '|g(x)-x| (Error Final)': 'N/A'
        })

# Crear el DataFrame de Pandas para la tabla
df_resultados = pd.DataFrame(resultados)
df_resultados
```

## Sistema de ecuaciones no lineales

\begin{align*}
3x_1 - \cos(x_1 x_2) - \frac{1}{2} & = 0 \\
x_1^{2} - 81(x_2+0.1)^{2} + \sin(x_3) + 1.06 & = 0 \\
\text{e}^{-x_1 x_2} + 20 x_3 + \frac{10\pi -3}{3} & = 0
\end{align*}

Para una primera iteracion de punto fijo proponemos:
\begin{align*}
g_1(x_1,x_2,x_3) & = \frac{\cos(x_1 x_2)}{3} + \frac{1}{6}\\
g_2(x_1,x_2,x_3) & = \frac{1}{9}\sqrt{x_1^{2} + \sin(x_3) + 1.06} - 0.1\\
g_3(x_1,x_2,x_3) & = -\frac{1}{20} \text{e}^{-x_1 x_2} - \frac{10\pi -3}{60}
\end{align*}

```python
from numpy import cos, sin, exp, pi, sqrt
import numpy as np
def F(x):
    f1 = 3*x[0] - cos(x[1]*x[2]) - 0.5 
    f2 = x[0]**2 - 81*(x[1]+0.1)**2 + sin(x[2]) + 1.06
    f3 = exp(-x[0]*x[1]) + 20*x[2] + (10*pi-3)/3
    return np.array([f1,f2,f3])
```

## Newton-Raphson
Aplicamos el m\'etodo de Newton. Calculamos una expresion de la matriz Jacobiana.

```python
def JF(x):
    f11 = 3
    f12 = x[2]*sin(x[1]*x[2])
    f13 = x[1]*sin(x[1]*x[2])
    f21 = 2*x[0]
    f22 = -162*(x[1]+0.1)
    f23 = cos(x[2])
    f31 = x[1]*exp(-x[0]*x[1])
    f32 = -x[0]*exp(-x[0]*x[1])
    f33 = 20
    return np.array([[f11,f12,f13],[f21,f22,f23],[f31,f32,f33]])
```

```python
MAX_ITER = 5
x = np.array([0.1,0.1,-0.1])
print( "  x0:", x)
print( "  F(x0):", F(x))
for k  in range(1,MAX_ITER+1):
    x = x - np.linalg.solve(JF(x),F(x))
    print( "* iteracion ", k)
    print( "  x:", x)
    print( "  F(x):", F(x))
    print("")
    
```

```python
def Newton(F, JF, x0, tol=1e-8, Maxiter=20):
    x = x0.copy()
    k = 1
    xlist = [x]
    while k <= Maxiter:
        xnew = x - np.linalg.solve(JF(x), F(x))
        xlist += [xnew]
        if np.linalg.norm(F(xnew))<tol: return xnew, xlist
        k += 1
        x = xnew.copy()
    print("se excedio el numero maximo de iteraciones")
    return xnew, xlist
```

## Convergencia Acelerada.

1.  Aceleracion de Aitken para una sucesion que converge linealmente.
    $$
\hat{p}_n = p_{n} - \frac{(p_{n+1}-p_{n})^{2}}{p_{n+2}-2p_{n+1} + p_{n}}
    $$
    **Ejemplo:** $$p_{n} = \cos(1/n), \quad \text{converge linealmente a } p =1$$

```python
p = lambda n: np.cos(1/n)
```

```python
nplot = np.arange(1,100)
pn = p(nplot)
fig, ax = plt.subplots(figsize=(4, 3))
ax.plot(nplot,pn)
fig, ax = plt.subplots(figsize=(4, 3))
ax.loglog(nplot,np.abs(p(nplot)-1.0))
plt.show()
```

```python
def aitken(pn):
    n = pn.size
    phat = np.zeros(n-2)
    for i in range(n-2):
        phat[i] = pn[i] - (pn[i+1]-pn[i])**2/(pn[i+2]-2*pn[i+1]+pn[i])
    return phat
phat = aitken(p(nplot))
```

```python
fig, ax = plt.subplots(figsize=(4, 3))
ax.loglog(nplot,np.abs(p(nplot)-1.0))
ax.loglog(nplot[2:],np.abs(phat-1.0))

plt.show()
```

```python
def Broyden(F, JFx0, x0, tol, Maxiter):
    A0inv = np.linalg.inv(JFx0)
    xsol = [x0]
    x1 = x0 - A0inv@F(x0)
    xsol += [x1]
    k = 2
    while k <=Maxiter:
        y1 = F(x1)-F(x0)
        s1 = x1-x0
        print(y1, s1)
        A1inv = A0inv + ( np.outer((s1-A0inv@y1), A0inv.T@s1))/(s1.dot(A0inv@y1))
        xnew = x1 - A1inv@F(x1)
        xsol += [xnew]
        if np.linalg.norm(F(xnew))<tol: return x1, xsol
        k += 1
        x0 = x1.copy()
        x1 = xnew.copy()
        A0inv = A1inv.copy()
    print("se excedio el numero maximo de iteraciones")
    return x1, xsol
        
    
```
