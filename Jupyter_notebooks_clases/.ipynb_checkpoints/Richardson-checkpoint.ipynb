# Idea

Muchos m&eacute;todos num&eacute;ricos calculan una aproximaci&oacute;n $A(h)$ de una cantidad desconocida $A$ (una derivada, una integral, ...) que depende de un par&aacute;metro $h>0$. Si conocemos la **estructura del error**, es decir, c&oacute;mo depende el error de $h$, podemos combinar dos aproximaciones $A(h)$ y $A(h/2)$ para **cancelar el t&eacute;rmino principal del error** y obtener una aproximaci&oacute;n de mayor orden, sin necesidad de deducir una f&oacute;rmula nueva desde cero.

Esta t&eacute;cnica se llama **extrapolaci&oacute;n de Richardson** (L. F. Richardson, 1911).

# Teor&iacute;a

Supongamos que la aproximaci&oacute;n $A(h)$ admite un desarrollo del error de la forma
$$
A(h) = A + c_1 h^{p_1} + c_2 h^{p_2} + c_3 h^{p_3} + \cdots, \qquad 0 < p_1 < p_2 < p_3 < \cdots,
$$
donde las constantes $c_1, c_2, \dots$ **no dependen de $h$** (pero son desconocidas). En particular $|A - A(h)| = \mathcal{O}(h^{p_1})$.

Evaluamos tambi&eacute;n con paso $h/2$:
$$
A(h/2) = A + c_1 \frac{h^{p_1}}{2^{p_1}} + c_2 \frac{h^{p_2}}{2^{p_2}} + \cdots
$$
Multiplicamos la segunda ecuaci&oacute;n por $2^{p_1}$ y le restamos la primera. El t&eacute;rmino $c_1h^{p_1}$ se cancela:
$$
2^{p_1}A(h/2) - A(h) = (2^{p_1}-1)A + c_2\big(2^{p_1-p_2}-1\big)h^{p_2} + \cdots
$$
Dividiendo por $2^{p_1}-1$ obtenemos la **f&oacute;rmula de extrapolaci&oacute;n de Richardson**:

<div style="background-color: #e3f2fd; border-left: 5px solid #1565c0; padding: 10px; border-radius: 5px;">

$$
R(h) := \frac{2^{p_1}A(h/2) - A(h)}{2^{p_1}-1} = A(h/2) + \frac{A(h/2)-A(h)}{2^{p_1}-1}, \qquad |A - R(h)| = \mathcal{O}(h^{p_2}).
$$

</div>

**Observaciones.**
<ul>
<li>Solo necesitamos conocer el exponente $p_1$, no la constante $c_1$.</li>
<li>La segunda forma muestra que $R(h)$ es $A(h/2)$ m&aacute;s una <i>correcci&oacute;n</i>. Adem&aacute;s, $\frac{A(h/2)-A(h)}{2^{p_1}-1}$ es una estimaci&oacute;n del error de $A(h/2)$.</li>
<li>En vez de $h/2$ se puede usar $h/t$ con $t>1$: $R(h) = \frac{t^{p_1}A(h/t)-A(h)}{t^{p_1}-1}$.</li>
<li>La ganancia de orden requiere que el desarrollo del error exista, es decir, que la funci&oacute;n sea suficientemente regular.</li>
</ul>

## Extrapolaci&oacute;n repetida: tabla de Richardson

Como $R(h)$ tiene a su vez un desarrollo del error (empezando en $h^{p_2}$), podemos volver a extrapolar para eliminar $c_2h^{p_2}$, luego $c_3h^{p_3}$, etc. Definimos la tabla
$$
T_{i,0} = A\big(h_0/2^{i}\big), \qquad
T_{i,k} = \frac{2^{p_k}\,T_{i,k-1} - T_{i-1,k-1}}{2^{p_k}-1}, \quad k = 1,\dots,i.
$$
La columna $k$ tiene error $\mathcal{O}(h^{p_{k+1}})$:

$$
\begin{array}{ccccc}
T_{0,0} & & & & \\
T_{1,0} & T_{1,1} & & & \\
T_{2,0} & T_{2,1} & T_{2,2} & & \\
T_{3,0} & T_{3,1} & T_{3,2} & T_{3,3} & \\
\vdots & \vdots & \vdots & \vdots & \ddots \\
\mathcal{O}(h^{p_1}) & \mathcal{O}(h^{p_2}) & \mathcal{O}(h^{p_3}) & \mathcal{O}(h^{p_4}) &
\end{array}
$$

*Nota:* aplicada a la regla del trapecio (con $p_k = 2k$) esta tabla es el **m&eacute;todo de Romberg** para integraci&oacute;n num&eacute;rica.

# Ejemplo: extrapolaci&oacute;n de la diferencia forward

Sea $f$ suficientemente regular y $D_+(h) := D_+(f,x_0,h) = \frac{f(x_0+h)-f(x_0)}{h}$. Por Taylor,
$$
f(x_0+h) = f(x_0) + hf'(x_0) + \frac{h^2}{2}f''(x_0) + \frac{h^3}{6}f'''(x_0) + \frac{h^4}{24}f^{(4)}(x_0) + \cdots
$$
y por lo tanto
$$
D_+(h) = f'(x_0) + \underbrace{\frac{f''(x_0)}{2}}_{c_1}h + \underbrace{\frac{f'''(x_0)}{6}}_{c_2}h^2 + \underbrace{\frac{f^{(4)}(x_0)}{24}}_{c_3}h^3 + \cdots
$$
Es decir, $A = f'(x_0)$ y los exponentes son $p_k = k$: $\;p_1 = 1,\ p_2 = 2,\ p_3 = 3, \dots$

**Primera extrapolaci&oacute;n** ($p_1 = 1$):
$$
R_1(h) = \frac{2D_+(h/2) - D_+(h)}{2-1} = 2D_+(h/2)-D_+(h) = \frac{-f(x_0+h) + 4f\big(x_0+\tfrac h2\big) - 3f(x_0)}{h},
\qquad |f'(x_0)-R_1(h)| = \mathcal{O}(h^2)\ \text{ si } f\in C^3.
$$
&iexcl;Obtenemos una f&oacute;rmula *unilateral* de segundo orden sin hacer el desarrollo de Taylor de la nueva f&oacute;rmula!

**Segunda extrapolaci&oacute;n** ($p_2 = 2$):
$$
R_2(h) = \frac{4R_1(h/2) - R_1(h)}{3}, \qquad |f'(x_0)-R_2(h)| = \mathcal{O}(h^3)\ \text{ si } f\in C^4.
$$

Verificamos esto num&eacute;ricamente con $f(x) = xe^x$, $x_0 = 1$, donde $f'(x) = (1+x)e^x$, es decir $f'(1) = 2e$.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['figure.figsize'] = (10, 6)
plt.rc('font', size=14)
```

```python
# Funcion y derivada exacta
f  = lambda x: x*np.exp(x)
fp = lambda x: (1 + x)*np.exp(x)
x0 = 1.0
exacto = fp(x0)          # f'(1) = 2e

# Diferencia forward: error O(h)
Dp = lambda h: (f(x0 + h) - f(x0)) / h

# Primera extrapolacion (p1 = 1): error O(h^2)
R1 = lambda h: 2*Dp(h/2) - Dp(h)

# Segunda extrapolacion (p2 = 2): error O(h^3)
R2 = lambda h: (4*R1(h/2) - R1(h)) / 3

# Sucesion de pasos h = 2^{-k}
h = 2.0 ** (-np.arange(1, 11))

err_Dp = np.abs(exacto - Dp(h))
err_R1 = np.abs(exacto - R1(h))
err_R2 = np.abs(exacto - R2(h))

# Grafico log-log
plt.figure()
plt.loglog(h, err_Dp, 'o-', label='$D_+(h)$', linewidth=3)
plt.loglog(h, err_R1, 's-', label='$R_1(h)$', linewidth=3)
plt.loglog(h, err_R2, '^-', label='$R_2(h)$', linewidth=3)
plt.loglog(h, 2*h,    '--k', label='$\\mathcal{O}(h)$')
plt.loglog(h, h**2,   ':k',  label='$\\mathcal{O}(h^2)$')
plt.loglog(h, h**3/4, '-.k', label='$\\mathcal{O}(h^3)$')
plt.xlabel('h'); plt.ylabel('error')
plt.legend(bbox_to_anchor=(1.05, 0.5), loc='center left')
plt.grid(True, which='both', ls='--')
plt.tight_layout()
plt.show()
```

## Orden de convergencia observado

Si $e(h) \approx C h^{\alpha}$, entonces $\dfrac{e(h)}{e(h/2)} \approx 2^{\alpha}$, de donde
$$
\alpha \approx \log_2\left(\frac{e(h)}{e(h/2)}\right).
$$

```python
# Orden observado: alpha = log2( e(h) / e(h/2) )
orden = lambda e: np.log2(e[:-1] / e[1:])

print(f"{'h':>10} {'err D+':>10} {'orden':>6} {'err R1':>10} {'orden':>6} {'err R2':>10} {'orden':>6}")
oDp, oR1, oR2 = orden(err_Dp), orden(err_R1), orden(err_R2)
for i in range(len(h)):
    o = lambda arr: f"{arr[i-1]:6.2f}" if i > 0 else f"{'-':>6}"
    print(f"{h[i]:10.2e} {err_Dp[i]:10.2e} {o(oDp)} {err_R1[i]:10.2e} {o(oR1)} {err_R2[i]:10.2e} {o(oR2)}")
```

## Tabla de Richardson completa

Implementamos la tabla $T_{i,k}$ de forma general: recibe la aproximaci&oacute;n $A(h)$, el paso inicial $h_0$, el n&uacute;mero de filas $n$ y la lista de exponentes $p_1, p_2, \dots$

```python
def tabla_richardson(A, h0, n, p):
    # A  : funcion h -> A(h)
    # h0 : paso inicial
    # n  : numero de filas (pasos h0, h0/2, ..., h0/2^{n-1})
    # p  : exponentes del desarrollo del error [p1, p2, ...]
    T = np.full((n, n), np.nan)
    for i in range(n):
        T[i, 0] = A(h0 / 2**i)                      # columna 0: aproximacion base
        for k in range(1, i + 1):
            t = 2.0 ** p[k-1]
            T[i, k] = (t*T[i, k-1] - T[i-1, k-1]) / (t - 1)   # elimina c_k h^{p_k}
    return T

n = 6
T = tabla_richardson(Dp, h0=0.5, n=n, p=[1, 2, 3, 4, 5])   # forward: p_k = k
E = np.abs(T - exacto)

np.set_printoptions(precision=2, linewidth=120)
print("Errores |f'(x0) - T[i,k]|:")
print(E)

print("\nOrden observado por columna:")
for k in range(4):
    print(f"columna {k}: ", np.round(orden(E[k:, k]), 2))
```

Cada columna gana un orden: $1, 2, 3, 4, \dots$ Con solo $6$ evaluaciones de $D_+$ (pasos $h_0, \dots, h_0/32$) obtenemos un error del orden de $10^{-9}$, mientras que la diferencia forward con el paso m&aacute;s fino tiene un error de $\approx 6\cdot 10^{-2}$.

# &iquest;Qu&eacute; pasa si $f$ no es suficientemente regular?

La extrapolaci&oacute;n supone que existe el desarrollo $A(h) = A + c_1h^{p_1} + c_2h^{p_2} + \cdots$. Si $f$ no es regular, este desarrollo no existe y **no se gana orden**. Por ejemplo, $f(x) = x^{2.5}$ para $x\geq0$, $f(x)=0$ para $x<0$, en $x_0 = 0$: $f\in C^2$ pero $f\notin C^3$. Aqu&iacute; $f'(0) = 0$ y
$$
D_+(h) = h^{1.5}, \qquad R_1(h) = 2\Big(\frac h2\Big)^{1.5} - h^{1.5} = \big(2^{-0.5}-1\big)h^{1.5},
$$
es decir, ambos convergen con orden $1.5$.

```python
g = lambda x: np.heaviside(x, 1) * np.abs(x)**2.5
x0g = 0.0

Dp_g = lambda h: (g(x0g + h) - g(x0g)) / h
R1_g = lambda h: 2*Dp_g(h/2) - Dp_g(h)

e_Dp_g = np.abs(0 - Dp_g(h))
e_R1_g = np.abs(0 - R1_g(h))

print("orden D+ :", np.round(orden(e_Dp_g), 2))
print("orden R1 :", np.round(orden(e_R1_g), 2))

plt.figure()
plt.loglog(h, e_Dp_g, 'o-', label='$D_+(h)$', linewidth=3)
plt.loglog(h, e_R1_g, 's-', label='$R_1(h)$', linewidth=3)
plt.loglog(h, h**1.5, '--k', label='$\\mathcal{O}(h^{1.5})$')
plt.xlabel('h'); plt.ylabel('error')
plt.legend(); plt.grid(True, which='both', ls='--')
plt.tight_layout(); plt.show()
```

# Comentarios finales

<ul>
<li>La extrapolaci&oacute;n de Richardson es una t&eacute;cnica <b>general</b>: sirve para derivadas, integrales (Romberg), soluciones de ecuaciones diferenciales, etc. Solo requiere conocer los exponentes $p_k$ del desarrollo del error.</li>
<li>Aplicada a la diferencia central, cuyo desarrollo contiene solo potencias pares ($p_k = 2k$), cada extrapolaci&oacute;n gana <b>dos</b> &oacute;rdenes.</li>
<li>La extrapolaci&oacute;n <b>no elimina</b> los problemas de estabilidad: cuando $h\to 0$ los errores de redondeo $\sim\varepsilon/h$ siguen apareciendo. La ventaja es que se alcanza una buena precisi&oacute;n con $h$ moderados, antes de que el redondeo domine.</li>
<li>Si $f$ no es suficientemente regular, el desarrollo del error no existe y la extrapolaci&oacute;n no mejora el orden.</li>
</ul>
